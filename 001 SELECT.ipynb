# SELECT y FROM
SELECT elige QUE columnas quieres ver. FROM dice DE DONDE. Es como decir 'dame el nombre y precio DE la
tabla productos'. Todo query empieza aqui. Puedes usar * para traer todo, pero en produccion siempre especifica
columnas.


### Muestra todas las columnas de la tabla actor

```python

%%sql
-- Muestra todas las columnas de la tabla actor
SELECT * 
FROM actor
LIMIT 5;
```

### Muestra solo el título y la duración de todas las peliculas

```python
%%sql
SELECT title AS Pelicula, length AS Duracion
FROM film
LIMIT 5;
```

### Puedes hacer operaciones y literales en las consultas SQL. Por ejemplo, puedes concatenar el nombre y apellido de los actores en una sola columna llamada NombreCompleto:   

```python
%%sql
SELECT CONCAT(first_name, ' ', last_name) AS NombreCompleto, 2+2 AS Suma, "Constante"AS Textos
FROM actor
LIMIT 10;
```

### la siguiente query muestra el uso de  literales, , una expresión, y una función built-in.


```python
%%sql
SELECT language_id,
    'COMUN' tipo_lenguaje,
    language_id * 3.1416  lang_pi_value,
    upper(name) Nombre_idioma
FROM language
LIMIT 5;
```

Removuiendo duplicados , en unos caso sera necesario obtener los valore sunicos de una columna, para esto se utiliza la palabra reservada DISTINCT, veamos cuantos actores haya en la tabla actorm

```python
%%sql 
DESC actor;
```

```python
%%sql
SELECT DISTINCT actor_id 
FROM actor
LIMIT 5;
```

```python
%%sql
-- Lista nombre y apellido de cada actor
SELECT first_name as Nombre,
    last_name as Apellido
FROM actor
LIMIT 10;
```

```python
%%sql
-- Muestra title, release_year y rating de film
SELECT title AS Pelicula, 
    release_year AS Año,
    rating AS Clasificación
FROM film 
LIMIT 5;
```

```python
%%sql
-- Ordena las peliculas por titulo de forma ascendente
SELECT title AS Pelicula, 
    length AS Duracion
FROM film
ORDER BY title ASC
LIMIT 10;
```

```python
%%sql 
-- Ordena las peliculas por duracion de forma descendente y luego por title ascendente
SELECT title AS pelicula, 
    length AS duracion
FROM film
ORDER BY duracion DESC, title ASC
LIMIT 10;
```

```python
%%sql 
-- Muestra las 15 peliculas mas largas
SELECT title AS pelicula, length AS duracion
FROM film
ORDER BY duracion DESC
LIMIT 15;
```

```python
%%sql
-- lista clientes con alias nombre_completo, concatenando first_name y last_name
SELECT CONCAT(first_name,' ',last_name) AS nombre_completo
FROM customer
LIMIT 10;
```

```python
%%sql 
-- muestra customer_id, email, y active con alias en español
SELECT customer_id AS id_cliente, email as correo_electronico, active AS activo
FROM customer
LIMIT 10;
```
