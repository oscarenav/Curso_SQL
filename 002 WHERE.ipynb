## WHERE en SQL

`WHERE` se usa para **filtrar filas** en una consulta: solo devuelve los registros que cumplen una condición.

**Sintaxis:**
```sql
SELECT columnas
FROM tabla
WHERE condicion;
```

**Ejemplo:**
```sql
SELECT title, length
FROM film
WHERE length > 120;
```

**Operadores comunes:** `=`, `!=`, `>`, `<`, `>=`, `<=`, `AND`, `OR`, `IN`, `BETWEEN`, `LIKE`, `IS NULL`.

**Idea clave:** `WHERE` filtra los datos antes de mostrarlos en el resultado.

#### Selecciona las peliculas  que tengan  rating G y una duración de renta mayo igual a 7


````python
%%sql
SELECT title
FROM film 
WHERE rating = 'G' AND rental_duration >= 7;
````

Ahora las peliculas que tengan rating 'G'y una duración de renta mayor o igual a 7, además de aquellas que tengan rating 'PG-13' y una duración de renta menor o iual a 5

````python
%%sql
DESC film;
SELECT title AS pelicula,
    rating  AS Clasificacion, 
    rental_duration AS Duracion_Alquiler
FROM film
WHERE (rating = 'G' and rental_duration >= 7) OR (rating = 'PG' and rental_duration >= 5)
ORDER BY Clasificacion DESC, Duracion_Alquiler DESC;
````

````python
%%sql 
-- Muestra peliculas con clasifficacion PG 
SELECT title as Pelicula
FROM film
WHERE rating = 'PG'
LIMIT 10;
````

````python
%%sql
DESC customer;
SELECT customer_id, first_name, last_name, email
FROM customer 
WHERE active = 1
LIMIT 10;
````

````python
%%sql
-- Muestra peliculas con rental_rate mayor a 2.99
SELECT title, round(rental_rate, 2)
FROM film
WHERE rental_rate > 2.99
ORDER BY rental_rate DESC;
````

````python
%%sql
-- Busca peliculas con length entre 90 y 120
SELECT title AS pelicula,
length AS duracion
FROM film
WHERE length BETWEEN 90 AND 120
LIMIT 10;
````
