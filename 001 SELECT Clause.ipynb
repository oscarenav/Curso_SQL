## Resumen: cláusula SELECT

La cláusula `SELECT` se usa para consultar datos de una o varias columnas en una tabla.

- `SELECT *` devuelve todas las columnas.
- `SELECT columna1, columna2` devuelve solo columnas específicas.
- `FROM` indica la tabla de origen.
- `WHERE` filtra filas según una condición.
- `ORDER BY` ordena los resultados (ASC o DESC).
- `LIMIT` restringe la cantidad de filas devueltas.

Ejemplo básico:

```sql
SELECT nombre, edad
FROM empleados
WHERE edad >= 18
ORDER BY edad DESC
LIMIT 5;
```
